```kotlin
@DataSchema
interface Matrx {
	val gameId: Int
	val gameState: GameState
	val seed: Long
	val timestamp: Long
	val turn: Int
	val winner: Winner

	@DataSchema(isOpen = false)
	interface CurrentPlayer {
		val capturedPieces: List<Int>
		val depth: Int
		val enableFPU: Boolean
		val enablePW: Boolean
		val iterations: Int
		val model: String
		val name: String
		val piecesInReserve: List<Int>
		val strength: String
		val timeControl: Boolean
		val timeDuration: String
		val useRAVE: Boolean
	}

	@DataSchema(isOpen = false)
	interface GameState {
		val currentPlayer: CurrentPlayer
		val nextPlayer: CurrentPlayer
		val turnDuration: TurnDuration
	}

	@DataSchema(isOpen = false)
	interface Winner {
		val loser: CurrentPlayer
		val winCondition: String
		val winner: CurrentPlayer
	}
}
```
